```python
# Broadcasting !!! Very Important
# Need Keepdim = True here because otherwise we would be broadcasting one row across all of the rows as 
# opposed to just having one row total vector (aka a column) being broadcast across all of the rows. 
P = (N+1).float()
P = P / P.sum(1, keepdim = True)
```

```python
#log (a+b+c) = log (a) + log(b) + log (c)
# therefore when taking the likelihood (aka the probability that the model outputs the exact dataset)
# but to make it more user friendly using log (likelihood) so that we have whole numbers instead of small percentages
# then adding up all of those values bc of the truth above. 
#that gives us a decent judge for how good this model is
```

```python

log_likelihood = 0.0
n=0 
# for w in words:
for w in ["nicholas"]:
    chs = ['.'] + list(w) + ['.']
    for ch1, ch2 in zip(chs,chs[1:]):
        ix1 = stoi[ch1]
        ix2 = stoi[ch2]
        prob = P[ix1, ix2]
        logprob = torch.log(prob) 
        log_likelihood += logprob
        n+=1
        print(f'{ch1}{ch2}:{prob:.4f} {logprob:.4f} {log_likelihood}')

nll = -log_likelihood
print(f'{nll=}') 
print(f'{nll/n=}')
```

```python
#random initialize the 27 neuron weights - each neuron recieves 27 inputs. (the character pairings)
# g is rand int, and then w is the weight. putting random ws in the 27 x 27 matrix. 
g = torch.Generator().manual_seed(2147483647)
W = torch.randn((27,27),generator = g, requires_grad = True)
```

```python
# FORWARD PASS (everything here is just multiplcation and addition - therefore we can differentiate
# aka we are able to back propogate this forward pass. 

# first encode the xs into an array with 1s at the correspoding positions and 0s everwhere else. 
xenc = F.one_hot(xs,num_classes = 27).float() 
logits = xenc @ W # multiply the encoded xs by the weights to get fake counts - log counts
# SOFTMAX FUNCTION below ( takes in any numbers and outputs probabilities) 
counts = logits.exp() # equivalent to the N matrix from above (aka the matrix with the counts of each bigram) 
probs = counts / counts.sum(1, keepdims = True) # normalize them to create probablities. 
 #so this is just the probability of each of those bigrams occuring
# interperet this as the likelihood of each of the 27 characters has of coming next 

#Loss Function
loss = -probs[torch.arange(5), ys].log().mean()
```

```python
print(loss.item())
```

```python
# backwards pass 
W.grad = None # set gradient to 0 
loss.backward()
```

```python
# update the weights
W.data += -0.1 * W.grad
```

```python
#ALl together now. 
# input data, create all of the bigrams, organize them in the matrix
xs, ys = [],[]
for w in words:
    chs = ['.'] + list(w) + ['.']
    for ch1, ch2 in zip(chs,chs[1:]):
        ix1 = stoi[ch1]
        ix2 = stoi[ch2]
        xs.append(ix1)
        ys.append(ix2)
xs = torch.tensor(xs)
ys = torch.tensor(ys)
num = xs.nelement()
#initialize random weights
g = torch.Generator().manual_seed(2147483647)
W = torch.randn((27,27),generator = g, requires_grad = True)
```

```python
#gradient descent (by batching) 
for k in range (10):
    # Forward Pass
    xenc = F.one_hot(xs,num_classes = 27).float() 
    logits = xenc @ W
    counts = logits.exp() 
    probs = counts / counts.sum(1, keepdims = True)
    loss = -probs[torch.arange(num), ys].log().mean()
    print(loss.item())
    #Backward pass
    W.grad = None
    loss.backward()
    #update weights
    W.data += -50 * W.grad
```

```python
#sampling 
g = torch.Generator().manual_seed(2147483647)
for i in range (10):
    out = []
    ix = 0 
    while True :
        
        #p = N[ix].float()
        #p = p / p.sum()
        xenc = F.one_hot(torch.tensor([ix]),num_classes = 27).float() 
        logits = xenc @ W
        counts = logits.exp() 
        p = counts / counts.sum(1, keepdims = True)
        
        ix = torch.multinomial(p, num_samples = 1, replacement = True, generator = g).item()
        out.append(itos[ix])
        if ix ==0:
            break
    print (''.join(out))
```
